### (参考) 内包形式によるリストの作成
長いリストは自動で作成したい。そのための便利な方法がある。

```python
data1 = [1, 2, 4, 8, 16, 32, 64, 128]  # 2のn乗を手入力していくのは大変
print(data1)

data2 = []
for i in range(8):
    data2.append(2**i)                 # for文を使えば、自動化できるが、行数が多い
print(data2)

data3 = [2**n for n in range(8)]       # 内包形式なら、同じことが1行でできる
print(data3)
```

## 2.9 データ型の比較表（総まとめ）


| データ型 | 変更可能（ミュータブル） | 反復可能（イテラブル） |  順序を持つ（シーケンス型）   | 
| ---| ---- | --- | --- |
| int | x | x | x | 
| float | x | x | x | 
| bool | x | x | x | 
| str | x | o | o | 
| list | o | o | o | 
| tuple | x | o | o | 
| dict | o | o | x | 
| set | o | o | x | | T TD | TD |D |

#### 以下では上の表が正しいことを確認してみる

```python
a = 1
id(a)     # idはここでは物理メモリ上の番地だと思っておく
```

```python
a = 2     # 新しい値で上書きしたのでは無く、新たな変数aを作成した（同じaという名前で変数を新たに作成した）
id(a)     # 整数はメモリを消費しないため、新しいインスタンスを作成しても計算効率は悪くならない。（再利用するメリットが無い）
```

```python
list1 = [2, 3, 5, 7]
id(list1)
```

```python
list1.append(11)
list1
```

```python
id(list1) # 要素は追加されたが、メモリ上のアドレスは変わっていない。リストは一般に大きいので再利用すると効率が良いため。
```

```python
list2 = list1
id(list2)
```

```python
list2.extend([13,17])    # list2のお尻に要素を付け足す。  # list2 += [13, 17] としても同じ
list1                    # list1が知らないうちに書き換えられている！！！　# 不思議に思う人は陶山本などで勉強しよう
```

```python
str1 = "make"
str1[0]
```

```python
str1[0] = "f"    # string型はimmutable（書き換え不能）であることを確認してみる
```

## 最小二乗法　～散布図の真ん中を通る直線を引いてみよう！～

```python
from sklearn.datasets import load_diabetes  # 使うライブラリをまとめて宣言しておく  # インストールするにはコンソールにて pip install scikit-learn
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

X = load_diabetes().data[:,2].reshape(-1,1)  # 今回はライブラリに付属するデータを利用する
Y = load_diabetes().target

model = LinearRegression()  # 線形回帰を行う
model.fit(X, Y)             # BMI値から糖尿病の進行を予測する
```

```python
# ナイーブにプロット（あまり美しいとは言えないかも）
plt.scatter(X, Y)
plt.plot(X, model.predict(X), color='red')  # 回帰直線を赤線で加える
```

```python
# より美しいプロットを試みる
plt.figure(figsize=(6, 5))                 # 縦横の比を指定する
plt.scatter(X,Y)
plt.plot(X, model.predict(X), color='red') # 回帰直線を赤線で加える
plt.xlabel('BMI'); plt.ylabel('Target')    # x軸とy軸にラベルを加える
txt = f'Linear Regression:  Y = {model.coef_[0]:.1f} X + {model.intercept_:.1f}'
plt.title(txt)                             # 表題として回帰直線の数式を加える
plt.show()
```
